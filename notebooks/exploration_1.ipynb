## Histogram of text length

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))
sns.histplot(df_parquet['text_length'], kde=True)
plt.title("Distribution of Email Body Length")
plt.show()

```

##Histogram of URL count

```python
plt.figure(figsize=(10,5))
sns.histplot(df_parquet['url_count'], kde=True)
plt.title("Distribution of URL Count")
plt.show()

```

## Histogram of Maximum URL length

```python
plt.figure(figsize=(10,5))
sns.histplot(df_parquet['url_length_max'], kde=True)
plt.title("Distribution of Maximum URL Length ")
plt.show()
```

## BIVARIATE EDA : Text length vs label

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df_parquet, x='label', y='text_length')
plt.title("Text Length vs Label")
plt.show()

```

##BIVARIATE EDA:URL count vs label

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df_parquet, x='label', y='url_count')
plt.title("URL Count vs Label")
plt.show()
```

##MULTIVARIATE EDA: Correlation matrix

```python
numeric_cols = df_parquet.select_dtypes(include='number')
corr = numeric_cols.corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

##OUTLIER CHECKS: URL length max

```python
plt.figure(figsize=(10,4))
sns.boxplot(data=df_parquet, x='url_length_max')
plt.title("URL Length Max — Outlier Check")
plt.show()
```

## OUTLIER CHECKS: Text length

```python
plt.figure(figsize=(10,4))
sns.boxplot(data=df_parquet, x='text_length')
plt.title("Text Length — Outlier Check")
plt.show()
```

##Summary

###
- The dataset contains 108,685 rows and 21 columns.
- It includes sender/receiver metadata, timestamps, subject lines, email body text, URL features, attachment features, language detection, and a phishing label.
- Data types are mostly object for text fields and float64/int64 for numerical features
- From UNIVARIATE EDA  and the histograms
, we understood that , Most emails have few URLs, but phishing emails sometimes have a lot .Phishing emails often contain very long, suspicious URLs.Phishing URLs often hide inside many subdomains.
- From BIVARIATE EDA : Text length vs label, we now know that
Phishing emails tend to be short.

- The boxplot for url_length_max shows that most emails contain short URLs, but there is a single extreme outlier representing a very long URL (over 66,000 characters). This pattern shows that the long email with long url is most probably a phishing email.
- Long emails are not errors they are real content. In the boxplot for text_length we see that while phishing emails are generally shorter, both classes contain long outliers.We should not remove them during cleaning.
- The receiver and sender domain shows some  have multiple receiver/sender domains
, these columns also helps detect suspicious patterns
